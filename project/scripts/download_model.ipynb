# Save the embedding model into `models/`

C4's first search stage turns text into vectors with a sentence-transformers model (`all-MiniLM-L6-v2`, about 90 MB).
This notebook downloads it once and saves it into `models/` (git-ignored), so the project then runs **offline** from its own folder.
`Embedder` loads `models/<name>` automatically when it exists.

**Needs:** internet, once.

```python
from pathlib import Path
import sys

# the notebook may be opened from scripts/ or from the project folder: find the project root either way
here = Path.cwd().resolve()
PROJECT_ROOT = next(p for p in [here, *here.parents] if (p / "agents").is_dir() and (p / "shared").is_dir())
sys.path.insert(0, str(PROJECT_ROOT))
print("project root:", PROJECT_ROOT)
```

## Settings

```python
NAME = "all-MiniLM-L6-v2"   # the model to save
TARGET = PROJECT_ROOT / "models" / NAME
```

## 1 · Download and save

```python
from sentence_transformers import SentenceTransformer

SentenceTransformer(NAME).save(str(TARGET))
print("saved to", TARGET)
```

## 2 · Check that the project loads it from `models/`
A vector of 384 numbers means it works.

```python
from shared.embeddings.embedder import Embedder

embedder = Embedder(NAME)
print("loading from:", embedder.model_name)
print("vector length:", len(embedder.embed_text("user logs in")))
```
